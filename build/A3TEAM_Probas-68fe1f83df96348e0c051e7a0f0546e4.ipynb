**Exemple 3**

Soit $X$ une variable aléatoire de fonction de répartition :

```{math}
:enumerated: false

F_{X}\left( x\right) =
\begin{cases}
1-\exp \left( -3x\right) & \text{si } x\geq 0 \\
0 & \text{sinon}
\end{cases}
```

1. Déterminer la densité $f_X$ de $X$.
2. Calculer $\mathbb{P}\left( -2<X\leq 0.5\right)$.

````{dropdown} Solution (cliquez pour afficher)

1. $F_X$ est une fonction de répartition, donc toute fonction positive coïncidant avec $F'_X$ est une densité de $X$ et en particulier la fonction $f_X$ définie par

```{math}
:enumerated: false

f_{X}\left( x\right) =
\begin{cases}
3\exp \left( -3x\right) & \text{si } x\geq 0 \\ 
0 & si x<0
\end{cases}
```

2. Puisque $F_X$ est la fonction de répartition de $X$,

```{math}
:enumerated: false

\mathbb{P}\left(-2 < X \leq 0.5\right) &= F_X(0.5) - F_X(-2) \\
&= 1 - \exp(-3 \times 0.5) - 0 \\
&= 1 - e^{-1.5} \\
&\simeq 0.78
```

Grâce à la question précédente on aurait aussi pu utiliser la densité de $X$ :

```{math}
:enumerated: false

\mathbb{P}\left(-2 < X \leq 0.5\right) &= \int_{-2}^{0.5} f(x)\, dx \\
&= \int_{0}^{0.5} 3\exp(-3x)\, dx \\
&= 1 - e^{-1.5} \\
&\simeq 0.78
```

**Exemple 3 (suite)**

Soit $X$ une variable aléatoire de densité $f$ définie par $f\left(x\right)=3\exp \left( -3x\right) 1_{\mathbb{R}_+}\left(x\right)$.

Déterminer la loi de $X^2$.

````{dropdown} Solution (cliquez pour afficher)

$X^2\left(\Omega\right)=\mathbb{R}_+$.

Soit $t\in \mathbb{R}_+$.

```{math}
:enumerated: false

\mathbb{P}\left(X^2\leq t\right)  &=\mathbb{P}\left(\left|X\right|\leq \sqrt{t}\right) \\
&=\mathbb{P}\left(0\leq X\leq \sqrt{t}\right) \\
&=\int_0^{\sqrt{t}}3\exp \left( -3x\right) dx  \\
&=1-\exp\left( -3\sqrt{t}\right)

```

d'où la fonction de répartition $F_{X^{2}}$ de $X^2$ :

```{math}
:enumerated: false

F_{X^{2}}\left( t\right) =
\begin{cases}
0 & \text{si } t<0 \\ 
1-\exp \left( -3\sqrt{t}\right) & \text{si }t\geq 0
\end{cases}
```

$F_{X^{2}}$ est continue sur $\mathbb{R}^*_+$ et continue sur $\mathbb{R}^*_-$.

De plus $\lim\limits_{t\rightarrow 0^-}F_{X^{2}}\left(t\right)=\lim\limits_{t\rightarrow 0^+}F_{X^{2}}\left(t\right)=0=F_{X^{2}}\left(0\right)$.

Par conséquent, $F_{X^{2}}$ est continue sur $\mathbb{R}$.

$F_{X^{2}}$ est de classe $\mathcal{C}^1(\mathbb{R}^*)$ donc $X^{2}$ est une variable aléatoire continue et a pour densité :

```{math}
:enumerated: false

f_{X^{2}}\left( t\right) =F_{X^{2}}^{\prime }\left( t\right) =
\begin{cases}
0 & \text{si } t<0 \\ 
\frac{3}{2\sqrt{t}}\exp \left( -3\sqrt{t}\right) & \text{si }t\geq 0
\end{cases}
```
````

## Moments d'une variable aléatoire réelle

**Exemple 3 (suite)**

Soit $X$ une variable aléatoire de densité $f$ définie par $f\left(x\right)=3\exp \left( -3x\right) 1_{\mathbb{R}_+}\left(x\right)$.

Calculer l'espérance de $X$.

````{dropdown} Solution (cliquez pour afficher)

$\int_{-\infty }^{+\infty }x f\left( x\right) dx$ est de m\^{e}me nature que $%
\int_{0}^{+\infty }x f\left( x\right) dx$ car $f$ est nulle sur $\mathbb{R}_{-}^{\ast }.$

Soit $b>0$  
```{math}
:enumerated: false

\int_{0}^{b}\left|xf\left( x\right) \right|dx &=\int_{0}^{b}xf\left( x\right) dx \\
&=\int_{0}^{b}3x\exp \left( -3x\right) dx\\
&=\left[ -x\exp \left( -3x\right) \right] _{0}^{b}+\int_{0}^{b}\exp \left(
-3x\right) dx \\
&=-b\exp \left( -3b\right) +\left[ -\frac{1}{3}\exp \left( -3x\right) %
\right] _{0}^{b} \\
&=\frac{1}{3}-b\exp \left( -3b\right) -\frac{1}{3}\exp \left( -3b\right) 
```

et $\lim\limits_{b\rightarrow +\infty }\frac{1}{3}-b\exp \left( -3b\right) -%
\frac{1}{3}\exp \left( -3b\right) =\frac{1}{3}$ donc $\int_{0}^{+\infty
}xf\left( x\right) dx$ converge absolument et vaut $\frac{1}{3}$.


Ainsi $\mathbb{E}\left( X\right) =\frac{1}{3}$.

````

`````python
import numpy as np
import matplotlib.pyplot as plt

# --- Définition de la fonction de densité ---
def f(x):
    return np.where(x >= 0, 3 * np.exp(-3 * x), 0)

# --- Création des points pour tracer la courbe ---
x = np.linspace(-0.5, 2, 1000)
y = f(x)

fig, ax = plt.subplots(figsize=(6, 6))

# Tracé de la courbe de densité
ax.plot(x, y, color='blue', linewidth=1.2)

# Ligne verticale à x=0 (saut de la fonction, de 0 à 3)
ax.plot([0, 0], [0, 3], color='blue', linewidth=1.2)

# Ligne pointillée rouge verticale à x = 1/3
ax.axvline(x=1/3, color='red', linestyle='--', linewidth=0.8)

# Titre du graphique (comme une annotation en haut, façon R)
ax.set_title(r"$f(x) = 3\exp(-3x)$", fontsize=10, fontweight='bold', pad=15)

# Labels des axes
ax.set_xlabel("x")
ax.set_ylabel("f(x)")

# Limites des axes
ax.set_xlim(-0.5, 2)
ax.set_ylim(0, 3.2)

# Graduations sur l'axe x, avec ajout du "1/3"
xticks = [-0.5, 0.0, 0.5, 1.0, 1.5, 2.0]
xtick_labels = ["-0.5", "0.0", "0.5", "1.0", "1.5", "2.0"]
ax.set_xticks(xticks)
ax.set_xticklabels(xtick_labels)

# Ajout du label "1/3" sous l'axe x, à la position x=1/3
ax.text(1/3, -0.18, "1/3", color='black', ha='center', va='top', fontsize=9)

# Graduations sur l'axe y
ax.set_yticks(np.arange(0, 3.5, 0.5))

plt.tight_layout()

plt.show()
`````

**Exemple 3 (suite)**

Soit $X$ une variable aléatoire de densité $f$ définie par $f\left(x\right)=3\exp \left( -3x\right) 1_{\mathbb{R}_+}\left(x\right)$.

Calculer l'espérance de $X^2$.

````{dropdown} Solution (cliquez pour afficher)

La fonction $\varphi:x \mapsto x^2$ est continue sur $\mathbb{R}$.

$\int_{-\infty }^{+\infty }x^2 f\left( x\right) dx$ est de même nature que $%
\int_{0}^{+\infty }x^2 f\left( x\right) dx$ car $f$ est nulle sur $\mathbb{R}_{-}^{\ast }.$

Soit $b>0$  
```{math}
:enumerated: false

\int_{0}^{b}x^{2}f\left( x\right) dx &=\int_{0}^{b}3x^{2}\exp \left(
-3x\right) dx \\
&=\left[ -x^{2}\exp \left( -3x\right) \right] _{0}^{b}+2\int_{0}^{b}x\exp
\left( -3x\right) dx \\
&=-b^{2}\exp \left( -3b\right) +2\left\{ \left[ -\frac{x}{3}\exp \left(
-3x\right) \right] _{0}^{b}-\frac{1}{3}\int_{0}^{b}\exp \left( -3x\right)
dx\right\}  \\
&=-b^{2}\exp \left( -3b\right) -\frac{2}{3}b\exp \left( -3b\right) -\frac{2%
}{3}\int_{0}^{b}\exp \left( -3x\right) dx \\
&=-b^{2}\exp \left( -3b\right) -\frac{2}{3}b\exp \left( -3b\right) -\frac{2%
}{3}\left[ -\frac{1}{3}\exp \left( -3x\right) \right] _{0}^{b} \\
&=-b^{2}\exp \left( -3b\right) -\frac{2}{3}b\exp \left( -3b\right) +\frac{2%
}{9}\left[ 1-\exp \left( -3b\right) \right]  \\
&=\frac{2}{9}-\frac{2}{3}be^{-3b}-b^{2}e^{-3b}-\frac{2}{9}e^{-3b}
```

et $\lim\limits_{b\rightarrow +\infty }\frac{2}{9}-\frac{2}{3}%
be^{-3b}-b^{2}e^{-3b}-\frac{2}{9}e^{-3b}=\frac{2}{9}$ donc $%
\int_{0}^{+\infty }x^{2}f\left( x\right) dx$ converge absolument et vaut $%
\frac{2}{9}$.

Ainsi $\mathbb{E}\left( X^{2}\right) =\frac{2}{9}$.
````

**Exemple 3 (suite)**

Soit $X$ une variable aléatoire de densité $f$ définie par $f\left(x\right)=3\exp \left( -3x\right) 1_{\mathbb{R}_+}\left(x\right)$.

Calculer la variance de $X$.

````{dropdown} Solution (cliquez pour afficher)

La variable aléatoire $X$ possède un moment d'ordre 2 donc une variance et d'après la formule de Huygens-Koenig, 

```{math}
:enumerated: false

\mathbb{V}\left( X\right)  &=\mathbb{E}X^{2}-\left( \mathbb{E}X\right) ^{2} \\
&=\frac{2}{9}-\left( \frac{1}{3}\right) ^{2} \\
&=\frac{1}{9}
```
````

**Exemple 3 (suite)**

Soient $n>1$ et $X_1,..., X_n$ des variables aléatoires indépendantes de densité $f$ définie par
$f\left(x\right)=3\exp \left( -3x\right) 1_{\mathbb{R}_+}\left(x\right)$.

Déterminer la loi de $M_n=\max \left\{X_1,..., X_n\right\}$.

````{dropdown} Solution (cliquez pour afficher)

On a vu précédemment que la fonction de répartition des $X_i$ est 

```{math}
:enumerated: false

F_{X_{i}}\left( x\right) =
\begin{cases}
1-\exp \left( -3x\right) & \text{si } x\geq 0 \\ 
0 & \text{sinon }
\end{cases}
```


Pour tout $x\in \mathbb{R}$, 


```{math}
:enumerated: false

F_{M_{n}}\left( x\right)  &=\mathbb{P}\left( M_{n}\leq x\right)  \\
&=F_{X_{1}}\left( x\right) \cdots F_{X_{n}}\left( x\right) 
```

Si $x<0$ alors $F_{M_{n}}\left( x\right) =0$ et si $x\geq 0$ 

```{math}
:enumerated: false

F_{M_{n}}\left( x\right) =\left( 1-\exp \left( -3x\right) \right) ^{n}

```

Puisque $F_{M_{n}}$ est de classe $\mathcal{C}^{1}$ sauf peut \^{e}tre en 0, $M_{n}$
est une variable al\'{e}atoire de densit\'{e}%

```{math}
:enumerated: false

f_{M_{n}}\left( x\right) =F_{M_{n}}^{\prime }\left( x\right) =3n\exp \left(
-3x\right) \left( 1-\exp \left( -3x\right) \right) ^{n-1}1_{\mathbb{R}_{+}}\left( x\right) 

```

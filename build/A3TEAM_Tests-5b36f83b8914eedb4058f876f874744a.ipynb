# **Tests d'hypothèses**

## Introduction

````{admonition} **Exemple 1**
:class: exemple

Un célèbre équipementier automobile fait travailler ses ingénieurs sur un nouveau procédé de fabrication, plus écologique, de joints toriques utilisés dans les boîtes de vitesse automatique. La dureté Shore A des joints doit cependant rester conforme à la norme, à savoir qu'elle doit être en moyenne égale à 50 (semi rigide). Cette valeur, notée $\mu_{0}$, est appelée "norme" ou "cible".

La dureté moyenne réelle des joints produits avec ce nouveau procédé, notée $\mu$, est inconnue. Pour connaître $\mu$ il faudrait mesurer la dureté Shore A de tous les joints produits, et en calculer la moyenne, ce qui est logistiquement et économiquement irréalisable.

Pour contrôler le nouveau procédé de fabrication, un technicien prélève $n=10$ joints et détermine sur cet échantillon leur dureté Shore A moyenne, notée $\overline{x}_n$.

Compte tenu des fluctuations d'échantillonnage, $\overline{x}_n$ peut être différente de la valeur nominale sans pour autant que cela prouve la défaillance du nouveau procédé de fabrication. Peut-on dire que le nouveau procédé est défaillant si l'on obtient $\overline{x}_n=52$ ou $\overline{x}_n=49$ ?

L'objectif est donc de déterminer à partir de quelles limites, $\overline{x}_n$ sera considérée comme suffisamment éloignée de 50, pour prendre la décision de rejeter le nouveau procédé de fabrication.
````

## Procédure d'un test

````{admonition} **Exemple 2**
:class: exemple

Les OGM sont-ils bons pour la santé ?


| Point de vue de l'écologiste | Point de vue d'une multinationale qui fabrique des OGM |
|---|---|
|$H_{0}$ = "les OGM ne sont pas bons"|$H_{0}$ = "les OGM sont bons"|

Si $\beta $ n'est pas calculé, le choix de $H_{0}$ n'est donc pas innocent.
````

## Formalisation des tests paramétriques

````{admonition} **Exemple 1 (suite)**
:class: exemple

La dureté Shore A d'un joint torique fabriqué avec le nouveau procédé est considérée comme une variable aléatoire $X$ de moyenne $\mu$. Le paramètre $\theta$ à tester est donc ici $\mu$. La valeur cible est $\mu_0=50$ donc l'hypothèse $H_{0}$ est : "$\mu=50$".

Si l'on considère que le nouveau procédé est défaillant lorsque la dureté moyenne est différente de $50$ alors l'hypothèse alternative sera 

$H_{1}$ : "$\mu\neq 50$" et on utilisera donc un test bilatéral.

Si l'on considère que le nouveau procédé est défaillant lorsque la dureté moyenne est inférieure à $50$ alors l'hypothèse alternative sera 

$H_{1}$ : "$\mu< 50$" et on utilisera donc un test bilatéral gauche.

Si l'on considère que le nouveau procédé est défaillant lorsque la dureté moyenne est supérieure à $50$ alors l'hypothèse alternative sera 

$H_{1}$ : "$\mu> 50$" et on utilisera donc un test bilatéral droit.
````

## Tests de conformité

### Test de conformité d'une moyenne avec un échantillon issu d'une loi normale

````{admonition} **Exemple 1 (suite)**
:class: exemple

Supposons que $X$, la dureté Shore A de chaque joint torique, suit une loi normale d'espérance $\mu$ et de variance $\sigma^2 =25$.

1. On cherche à savoir si la dureté moyenne des joints est conforme au cahier des charges. Pour cela un technicien prélève un 
échantillon de $n=10$ joints toriques et mesure leur dureté.

    a. Déterminer les hypothèses à tester.
   
    b. Déterminer une statistique pivotale et sa loi si $H_0$ est vraie.
   
    c. Déterminer la région de rejet du test utilisé, au seuil de risque de 5\%.
   
    d. Le technicien obtient sur l'échantillon de 10 joints $\overline{x}_{10}=52$.
   
    Conclure au seuil de risque de 5\%.
3. Si le nouveau procédé de fabrication produit des joints dont la dureté moyenne réelle $\mu$ est de 55, quelle est la probabilité que l'on s'en aperçoive à partir d'un échantillon de taille 10 ? Comment s'appelle ce risque ?
4. Après la mise sur le marché des joints, des retours clients laissent penser que la dureté Shore A est trop importante. Le responsable du bureau d'études décide donc de faire de nouveaux tests statistiques, au seuil de risque de 5\%. Il prélève un échantillon de 50 joints toriques qui donnent une dureté Shore A moyenne de 53. 
Que conclure ?
````

````{dropdown} Solution (cliquez pour afficher)

1.
    a. On cherche à tester l'hypothèse $H_{0}$ : "$\mu=50$" contre $H_{1}$ : "$\mu\neq 50$" au seuil de risque $\alpha =0,05$.
    
    b. $\sigma^2$  est connue et $X_i\hookrightarrow \mathcal{N}\left(\mu;\sigma^2\right)$.   
       Pour effectuer le test on utilise la statistique pivotale $T_{10}=\frac{\overline{X}_n-\mu_{0}}{\frac{\sigma }{\sqrt{n}}}=\frac{\overline{X}_{10}-50}{\frac{5}{\sqrt{10}}}$ qui suit une loi $\mathcal{N}\left(0;1\right)$ si $H_0$ est vraie.

    c. La région de rejet est, au seuil de 0,05, est $\mathcal{R}_{0,05}=\left] -\infty ;-1,96\right[ \cup \left]1,96;+\infty \right[ $.
    
    d. $t_{10}=\frac{52-50}{\frac{5}{\sqrt{10}}}=1,26\notin \mathcal{R}_{0,05}$ donc on ne peut pas rejeter
l'hypothèse au seuil 0,05.

    Autrement dit, avec un risque de 5\%, on ne peut pas considérer que la norme n'est pas respectée.

    Pour calculer la $p$-value recherchons dans la table la valeur $F_{\mathcal{N}\left( 0;1\right)}\left(1,26\right)$. On obtient $0,8962$ qui correspond à $1-\frac{\alpha}{2}$ puisque le test est bilatéral. Ainsi, $\alpha= 2\left(1-0,8962\right)=0.2076$.

    La $p$-value vaut donc $20,76\%$, ce qui signifie que pour considérer que la norme n'est pas respectée il faut accepter un risque de $20,76\%$.
2. Le risque de 2ème espèce est donné par

```{math}
:enumerated: false

\begin{align*}
\beta & =\mathbb{P}_\mu\left( \left. \text{accepter }H_{0}\right\vert H_{0}\text{ fausse}\right) \\
& =\mathbb{P}_\mu\left( \left. 50-1,96\frac{5 }{\sqrt{10}}\leq \overline{X}_{10}\leq 50+1,96\frac{5}{\sqrt{10}}\right\vert \mu=55\right)\\
& =\mathbb{P}_\mu\left( \left. 46,9\leq \overline{X}_{10}\leq 53,0\right\vert \mu=55\right)
\end{align*}
```

Or sous l'hypothèse $\mu=55$, $\overline{X}_{10}\hookrightarrow \mathcal{N}\left(55;\frac{5}{\sqrt{10}}\right) $ donc 

```{math}
:enumerated: false

\begin{align*}
\beta & =\mathbb{P}_\mu\left(\frac{46,9-55}{\frac{5}{\sqrt{10}}}\leq \frac{%
\overline{X}_{10}-55}{\frac{5}{\sqrt{10}}}\leq \frac{53,09-55}{\frac{5}{%
\sqrt{10}}}\right) \\
& =F_{\mathcal{N}\left( 0;1\right) }\left( -5,12\right) -F_{\mathcal{N}%
\left( 0;1\right) }\left( -1,21\right) \\
& \simeq 1-0,8869=0,1131
\end{align*}
```

donc il y a un risque de 11,31 chances sur 100 de ne pas d\'{e}tecter le déréglage si la machine est réglée à 55 Shore A.

3. 
    - On cherche à tester l'hypothèse $H_{0}$ : "$\mu=50$" contre $H_{1}$ : "$\mu>50$" au seuil de risque $\alpha =0,05$.
$\sigma^2$ est connue et $X_i\hookrightarrow \mathcal{N}\left(\mu;\sigma^2\right)$.
    - Pour effectuer le test on utilise la statistique pivotale $T_{10}=\frac{\overline{X}_n-\mu_{0}}{\frac{\sigma }{\sqrt{n}}}=\frac{\overline{X}_{10}-50}{\frac{5}{\sqrt{10}}}$ qui suit une loi $\mathcal{N}\left(0;1\right)$ si $H_0$ est vraie.
    - La région de rejet est, au seuil de 0,05, est $\mathcal{R}_{0,05}=\left]1,65;+\infty \right[ $.
    - $t_{10}=\frac{53-50}{\frac{5}{\sqrt{10}}}=1,90\in \mathcal{R}_{0,05}$ donc on peut rejeter
l'hypothèse au seuil 0,05.
    Autrement dit, avec un risque de 5\%, on  peut considérer que l'outil est usé. 
````

````{admonition} **Exemple 1 (suite)**
:class: exemple

Supposons que $X$, la dureté Shore A  de chaque joint torique, suit une loi normale d'espérance $\mu$ et de variance $\sigma^2$.

On cherche à détecter si la dureté moyenne Shore A est conforme au cahier des charges. Pour cela un technicien prélève un 
échantillon de 10 joints toriques et mesure leur dureté.

1. Déterminer les hypothèses à tester.
2. Déterminer une statistique pivotale et sa loi si $H_0$ est vraie.
3. Déterminer la région de rejet du test utilisé, au seuil de risque de 5\%.
4. Le technicien obtient sur l'échantillon de 10 joints $\overline{x}_{10}=52$ et $s_{10}^2=20$.

Conclure au seuil de risque de 5\%.
````

    Pour effectuer le test on utilise la statistique pivotale $T_{10}=\frac{\overline{X}_n-\mu_{0}}{\frac{S_{n,c}}{\sqrt{n}}}=\frac{\overline{X}_n-50}{\frac{S_{n,c}}{\sqrt{10}}}$ qui suit une loi $Student\left(10-1\right)$ si $H_0$ est vraie.

3. La région de rejet est, au seuil de 0,05, est $\mathcal{R}_{0,05}=\left] -\infty ;-t_{9;0,975}\right[ \cup \left]t_{9;0,975};+\infty \right[=\left] -\infty ;-2,262\right[ \cup \left]2,262;+\infty \right[ $.

4. $s_{10,c}^2=\frac{10}{9}s_{10}^2\cong 22,22$ donc $s_{10,c}\cong\sqrt{22,22}$

    $t_{10}\cong \frac{52-50}{\frac{\sqrt{22,22}}{\sqrt{10}}}\cong 1,34\notin \mathcal{R}_{0,05}$ donc on ne peut pas rejeter l'hypothèse au seuil 0,05.
    
    Autrement dit, avec un risque de 5\%, on ne peut pas considérer que la dureté Shore A moyenne du nouveau procédé ne respecte pas la norme.

````{dropdown} Solution (cliquez pour afficher)

1. On cherche à tester l'hypothèse $H_{0}$ : "$\sigma^2=25$" contre $H_{1}$ : "$\sigma^2\neq 25$" au seuil de risque $\alpha =0,05$.

2. $\mu$ est inconnue et $X_i\hookrightarrow \mathcal{N}\left(\mu;\sigma^2\right)$.

    Pour effectuer le test on utilise la statistique pivotale $T_{10}=\frac{\left(n-1\right)S_{n,c}^2}{\sigma^2_0}=\frac{\left(20-1\right)S_{n,c}^2}{25}$ qui suit une loi $\chi^2\left(20-1\right)$ si $H_0$ est vraie.

3. La région de rejet, au seuil de 0,05, est

    $\mathcal{R}_{0,05} = \left[ 0; \chi^2_{19;0,025} \right[ \cup \left] \chi^2_{19;0,975}; +\infty \right[ = \left[ 0; 8{,}91 \right[ \cup \left] 32{,}85; +\infty \right[$

4. $s_{20,c}^2=\frac{20}{19}s_{20}^2\cong 21,05$.

    $t_{20}\cong \frac{\left(20-1\right)21,05}{25}\cong 16,00\notin \mathcal{R}_{0,05}$ donc on ne peut pas rejeter
l'hypothèse $H_0$ au seuil 0,05.

    Autrement dit, avec un risque de 5\%, on ne peut pas considérer que le nouveau process de fabrication produit des joints dont la variance de la dureté est différente de celle du process précédent.

### Test de conformité d'une proportion

````{admonition} **Exemple 3**
:class: exemple

Une ligne de production est considérée opérationnelle si sa fabrication ne contient pas plus de 6\% de pièces
défectueuses.
On prélève au hasard dans cette fabrication un échantillon de 200 pièces. La fabrication étant supposée de grande taille, on admet que cet échantillon est un échantillon aléatoire simple.

1. Déterminer les hypothèses à tester.
2. Déterminer une statistique pivotale et sa loi si $H_0$ est vraie.
3. Déterminer la région de rejet du test utilisé, au seuil de risque de 5\%.

4. L'échantillon fournit 17 pièces défectueuses.

Est-il possible, au seuil de risque 0,05, de conclure que la machine n'est pas opérationnelle ?
````

````{dropdown} Solution (cliquez pour afficher)

1. On cherche à tester l'hypothèse $H_{0}$ : "$p=0,06$" contre $H_{1}$ : "$p> 0,06$" au seuil de risque $\alpha =0,05$.
2. Notons $X_i$ la variable qui vaut 1 si la ième pièce de l'échantillon est défectueuse et 0 sinon.

    $X_i\hookrightarrow \mathcal{B}\left(p\right)$.

    $n \times 0,06 = 12>5$ et $n \times (1-0,06) = 188>5$ donc pour effectuer le test on utilise la statistique pivotale $T_{200}=\frac{F_{200}-0,06}{\sqrt{\frac{0,06\left(1-0,06\right)}{200}}}$ qui suit à peu près une loi $\mathcal{N}\left( 0;1\right)$ si $H_0$ est vraie.

3. La région de rejet, au seuil de 0,05, est 

    $\mathcal{R}_{0,05}=\left] u_{0,95};+\infty \right[= \left] 1,65;+\infty \right[$
4. $f_{200}=\frac{17}{200}=0,085$.

    $t_{200}=\frac{0,085-0,06}{\sqrt{\frac{0,06\left(1-0,06\right)}{200}}}\cong 1,49 \notin \mathcal{R}_{0,05}$ donc on ne peut pas rejeter l'hypothèse $H_0$ au seuil 0,05.
    
    Autrement dit, avec un risque de 5\%, on ne peut pas considérer que la machine n'est pas opérationnelle.
